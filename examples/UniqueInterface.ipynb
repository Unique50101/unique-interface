# Unique Interface quickstart
Run the cells in order in Colab, JupyterLab, or the VS Code notebook editor. This example runs a local Python callback, checks its API, and displays an inline chat.

```python
from unique_interface import Interface, Client, ChatInterface
from importlib.metadata import version

def greet(name):
    return f"Hello, {name}!"

demo = Interface(greet, title="Greeting")
app = demo.launch(block=False, open_browser=False)
try:
    response = Client(app.local_url).predict("Unique")
    assert response["outputs"] == ["Hello, Unique!"]
    print("Unique Interface", version("unique-interface"), "passed:", response["outputs"][0])
finally:
    app.close()
```

```python
try:
    from google.colab import output
except ImportError:
    pass
else:
    output.enable_custom_widget_manager()

def reply(message, history):
    return greet(message)

chat = ChatInterface(reply, title="Unique notebook chat").launch()
```
